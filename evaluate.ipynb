```python
from search import faiss_search, fts_search, bm25_search, hybrid_search, reciprocal_rank_fusion
queries = [
    # Chunk 248: project completion time & cost
    {"query": "project completion time calculation", "relevant_doc_ids": [248]},
    {"query": "relationship between project cost and completion time", "relevant_doc_ids": [248]},
    {"query": "how to measure completion time for projects", "relevant_doc_ids": [248]},
    {"query": "distribution of project cost and time", "relevant_doc_ids": [248]},

    # Chunk 13: Veo-3 and visual reasoning
    {"query": "Veo-3 visual trace reasoning performance", "relevant_doc_ids": [13]},
    {"query": "temporal coherence in video reasoning tasks", "relevant_doc_ids": [13]},
    {"query": "maze navigation evaluation criteria", "relevant_doc_ids": [13]},
    {"query": "multi-step object manipulation assessment", "relevant_doc_ids": [13]},

    # Chunk 56: evaluation criteria and scoring
    {"query": "instruction alignment scoring for video tasks", "relevant_doc_ids": [56]},
    {"query": "temporal consistency evaluation", "relevant_doc_ids": [56]}
]



```

```python
def hit_in_results(results, relevant_ids):
    for r in results:
        doc_id = r.get("doc_id") or (r["docmeta"][0] if r.get("docmeta") else None)
        if doc_id in relevant_ids:
            return True
    return False
```

```python
metrics = {"vector":0, "keyword":0, "hybrid":0}
k = 10
for qobj in queries:
    q = qobj["query"]
    rel = qobj["relevant_doc_ids"]
    vec = faiss_search(q, k=k)
    key = bm25_search(q, k=k)  # or fts_search
    hyb = hybrid_search(q, k=k, alpha=0.6, use_bm25=True)["merged"]
    if hit_in_results(vec, rel): metrics["vector"] += 1
    if hit_in_results(key, rel): metrics["keyword"] += 1
    if hit_in_results(hyb, rel): metrics["hybrid"] += 1

n = len(queries)
print("Recall@k vector:", metrics["vector"]/n)
print("Recall@k keyword:", metrics["keyword"]/n)
print("Recall@k hybrid:", metrics["hybrid"]/n)
```

```python
res = hybrid_search("how to calculate lead time", k=3, alpha=0.6, use_bm25=True)
for r in res["merged"]:
    print(r["doc_id"], r["score"], r["content"][:50])
```
